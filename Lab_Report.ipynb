## Task 3: Estimation

In this task, I will use the attention score dataset to estimate the mean attention score under three experimental conditions and a 95% confidence interval. The population is the attention score of the experimental conditions I will be focusing on, which includes length stimuli (L1), SNRdB (-6) and semanticity (0).

The results show a sample size of 25 across all three conditions. The mean attention score is 13.02 and 14.02 for the standard deviation, showing a low performance in attention scores. The confidence interval indicates that there is a 95% certainty that the average population is between 7.23 and 18.81.

````python
scores = att[
    (att["Semanticity"] == 0) &
    (att["LengthStim"] == "L1") &
    (att["SNRdB"] == -6)
]["AttentionScore"].dropna()

n = len(scores)

mean_score = scores.mean()
std_score = scores.std(ddof=1)

se = std_score / np.sqrt(n)

ci = stats.t.interval(0.95, df=n-1, loc=mean_score, scale=se)

print("Sample size:", n)
print("Mean Attention Score:", mean_score)
print("Standard Deviation:", std_score)
print("95% Confidence Interval:", ci)
````

## Task 4: Compare two experimental conditions
Null: No difference between noisy and noiseless environment in attention scores.

Alternative: A difference between noisy and noiseless environment in attention scores.

In this task, I will use the attention score dataset to compare E1 noisy environment (SNRdB: -6) and E2 noiseless environment (SNRdB: 1000) as well length stimuli (L1) and semanticity (0). This is a paired setting. The Shapiro test (p value 0.17) was used to do a normality assumption. The p value for both Student t test and Wilcoxon test where both below 0.001. The results below show that attention is increased in noiseless environment rather than noisy.

````python
D = pd.read_csv('https://nikeshbajaj.github.io/PhyaatDataset/PhyAAt_AttentionScoreData_v1.csv')

df_e1 = D[(D['SNRdB'] == -6) & (D['LengthStim'] == 'L1') & (D['Semanticity'] == 0)]
df_e2 = D[(D['SNRdB'] == 1000) & (D['LengthStim'] == 'L1') & (D['Semanticity'] == 0)]

df_e1.head()
df_e2.head()

np.c_[df_e1['SID'], df_e2['SID']]

x1 = df_e1['AttentionScore'].to_numpy()
x2 = df_e2['AttentionScore'].to_numpy()

plt.boxplot([x1, x2], tick_labels=['E1', 'E2'])
plt.ylabel('Attention Level')
plt.title('Two groups: Paired')
plt.show()

t, p = scipy.stats.shapiro(x1-x2)
print('test statistics :', t)
print('p-value : \t', p)

t, p = scipy.stats.ttest_rel(x1, x2, alternative='two-sided')
print('test statistics :', t)
print('p-value : \t', p)

t, p = scipy.stats.wilcoxon(x1, x2, alternative='two-sided')
print('test statistics :', t)
print('p-value : \t', p)
````

The Shapiro wilk test (p value 0.00) for both subjects, showing non normality. The non parametric test, Mann Whitney u, (p value of 0.08) and the parametric test, student t test, (p value of 0.14).
````python
s1 = "S11"
s2 = "S4"

group1 = att.loc[att["SID"] == s1, "AttentionScore"].dropna()
group2 = att.loc[att["SID"] == s2, "AttentionScore"].dropna()

x1 = group1.to_numpy()
x2 = group2.to_numpy()

n1 = len(x1)
n2 = len(x2)

mean1 = np.sum(x1) / n1
mean2 = np.sum(x2) / n2

print(f"{s1} Mean:", mean1)
print(f"{s2} Mean:", mean2)

print("\nShapiro Wilk test:")
t1, p1 = stats.shapiro(x1)
print("S11 test statistic:", t1)
print("S11 p-value:", p1)

t2, p2 = stats.shapiro(x2)
print("S4 test statistic:", t2)
print("S4 p-value:", p2)

print("\nMann Whitney U test:")
t, p = stats.mannwhitneyu(x1, x2, alternative='two-sided')
print('test statistics :', t)
print('p-value : \t', p)

print("\nStudent t test:")

t, p = stats.ttest_ind(x1, x2, equal_var=False)
print('test statistics :', t)
print('p-value : \t', p)
````
